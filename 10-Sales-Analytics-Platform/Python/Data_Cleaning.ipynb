```python
customers["CustomerAge"]=(
    customers["DateFirstPurchase"].dt.year - 
    customers["BirthDate"].dt.year
    )
```

```python
customers["CustomerAge"].describe()
```

```python
bins = [18, 25, 35, 45, 55, 100]

labels = [
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56+"
]

customers["AgeGroup"] = pd.cut(
    customers["CustomerAge"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
customers["AgeGroup"].value_counts().sort_index()
```

```python
bins=[0,30000,70000,float("inf")]
labels=["Low","Medium","High"]
customers["IncomeSegment"]= pd.cut(
customers["YearlyIncome"], bins=bins, labels=labels, include_lowest=True
)
```

```python
customers["IncomeSegment"].describe()
```
